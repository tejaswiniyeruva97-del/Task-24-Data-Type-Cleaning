```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Employee_ID': ['101', '102', '103', '104', '105'],
    'Name': ['Ravi', 'Priya', 'Arun', 'Sneha', 'Kiran'],
    'Age': ['22', '23', '25', '24', '26'],
    'Salary': ['25000', '30000', '35000', '28000', '40000'],
    'Joining_Date': [
        '2024-01-10',
        '2024-02-15',
        '2024-03-20',
        '2024-04-12',
        '2024-05-18'
    ],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'HR']
}

df = pd.DataFrame(data)

df
```

```python
df.head()
```

```python
df.info()
```

```python
print("Original Data Types:")
print(df.dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df['Age'] = df['Age'].astype(int)

print(df['Age'].dtype)
```

```python
df['Salary'] = pd.to_numeric(df['Salary'])

print(df['Salary'].dtype)
```

```python
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'])

print(df['Joining_Date'].dtype)
```

```python
df['Department'] = df['Department'].astype('category')

print(df['Department'].dtype)
```

```python
print("Corrected Data Types:")
print(df.dtypes)
```

```python
comparison = pd.DataFrame({
    'Original': [
        'object',
        'object',
        'object',
        'object',
        'object',
        'object'
    ],
    'Corrected': df.dtypes.astype(str)
}, index=df.columns)

comparison
```

```python
df.info()

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
df.to_csv('cleaned_employee_dataset.csv', index=False)

print("Cleaned dataset saved successfully!")
```
